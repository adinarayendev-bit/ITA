## Load and Unzip Dataset

### Subtask:
Load the zipped `KDDTest+.arff.zip` file, unzip it, and then load the `.arff` file into a pandas DataFrame. Inspect the first few rows and check for basic information like data types and missing values.


```python
import zipfile
import pandas as pd
from scipy.io import arff
import os
import io
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
from sklearn.metrics import pairwise_distances_argmin_min
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, accuracy_score
from sklearn.metrics import precision_recall_curve, PrecisionRecallDisplay
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from sklearn.ensemble import RandomForestClassifier
import tensorflow as tf
from tensorflow import keras
from sklearn.preprocessing import LabelEncoder
from keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras import Input


# Specify the path to the zipped dataset
zip_file_path = '/content/KDDTest+.arff.zip'

# Create a directory to extract the files into
extract_dir = 'extracted_data'
os.makedirs(extract_dir, exist_ok=True)

print("Libraries imported and paths defined.")

# Unzip the file
with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)

# Get the name of the unzipped .arff file
unzipped_files = os.listdir(extract_dir)
arff_file_name = [f for f in unzipped_files if f.endswith('.arff')][0]
arff_file_path = os.path.join(extract_dir, arff_file_name)

# --- Manual ARFF parsing to handle scipy.io.arff strictness ---
data_lines = []
column_names = []
in_data_section = False

with open(arff_file_path, 'r') as f:
    for line in f:
        line = line.strip()
        if not line or line.startswith('%'):
            continue
        if line.upper().startswith('@ATTRIBUTE'):
            parts = line.split()
            col_name = parts[1]
            column_names.append(col_name)
        elif line.upper().startswith('@DATA'):
            in_data_section = True
        elif in_data_section:
            data_lines.append(line)

# Create a StringIO object from the collected data lines
data_string_io = io.StringIO("\n".join(data_lines))

# Read the data into a pandas DataFrame using read_csv
df = pd.DataFrame()
if column_names and data_lines:
    df = pd.read_csv(data_string_io, header=None, names=column_names)

# Clean up object columns: strip whitespace and remove any single quotes
for col in df.select_dtypes(['object']).columns:
    df[col] = df[col].astype(str).str.strip().str.replace("'", "")

print("\nFirst 5 rows of the DataFrame:")
print(df.head())

print("\nDataFrame Info:")
df.info()

print("\nMissing values per column:")
print(df.isnull().sum())

# Separate features (X) and target (y)
X = df.drop(columns=["'class'"])
y = df["'class'"]

print("Features (X) and target (y) separated.")
print(f"Shape of X: {X.shape}")
print(f"Shape of y: {y.shape}")
print("First 5 rows of X:")
print(X.head())
print("First 5 rows of y:")
print(y.head())

# Identify categorical and numerical columns
categorical_cols = ["'protocol_type'", "'service'", "'flag'"]
numerical_cols = X.select_dtypes(include=['int64', 'float64']).columns.tolist()

# Remove categorical columns from numerical_cols if any overlap (shouldn't happen with select_dtypes but for safety)
numerical_cols = [col for col in numerical_cols if col not in categorical_cols]

# One-hot encode categorical features
X_categorical_encoded = pd.get_dummies(X[categorical_cols], columns=categorical_cols, drop_first=True)

# Scale numerical features
scaler = StandardScaler()
X_numerical_scaled = scaler.fit_transform(X[numerical_cols])
X_numerical_scaled = pd.DataFrame(X_numerical_scaled, columns=numerical_cols, index=X.index)

# Concatenate the preprocessed features
X_processed = pd.concat([X_numerical_scaled, X_categorical_encoded], axis=1)

print("Categorical columns encoded and numerical columns scaled.")
print(f"Shape of X_processed: {X_processed.shape}")
print("First 5 rows of X_processed:")
print(X_processed.head())

# Determine an appropriate number of clusters (k) using the elbow method
sum_of_squared_distances = []
K = range(1, 11)

for k in K:
    kmeans = KMeans(n_clusters=k, n_init='auto', random_state=42)
    kmeans.fit(X_processed)
    sum_of_squared_distances.append(kmeans.inertia_)

print("Inertia calculated for K values:")
for k_val, inertia_val in zip(K, sum_of_squared_distances):
    print(f"K={k_val}: Inertia={inertia_val:.2f}")

plt.figure(figsize=(10, 6))
plt.plot(K, sum_of_squared_distances, 'bx-')
plt.xlabel('k')
plt.ylabel('Sum of squared distances (Inertia)')
plt.title('Elbow Method For Optimal k')
plt.xticks(K)
plt.grid(True)
plt.show()

optimal_k = 5

kmeans = KMeans(n_clusters=optimal_k, n_init='auto', random_state=42)
kmeans.fit(X_processed)

X_processed['kmeans_cluster'] = kmeans.labels_

print(f"KMeans model fitted with {optimal_k} clusters.")
print("Distribution of samples across clusters:")
print(X_processed['kmeans_cluster'].value_counts().sort_index())

# Get cluster centroids
centroids = kmeans.cluster_centers_

# Calculate the distance of each sample to its assigned centroid
distances = kmeans.transform(X_processed.drop(columns=['kmeans_cluster']))
min_distances = np.min(distances, axis=1)

# Identify clusters with very few members (potential anomalies)
cluster_counts = X_processed['kmeans_cluster'].value_counts().sort_index()
small_clusters = cluster_counts[cluster_counts < 10].index.tolist()

print(f"\nSmall clusters identified (fewer than 10 members): {small_clusters}")

# Optionally, identify points that are far from their own centroids in larger clusters
threshold = np.percentile(min_distances, 99)

anomaly_indices_by_distance = X_processed.index[min_distances > threshold].tolist()

print(f"\nNumber of potential anomalies identified by distance from centroid (top 1%): {len(anomaly_indices_by_distance)}")
print("Indices of a few potential anomalies by distance:")
print(anomaly_indices_by_distance[:10])

print("\nFurther analysis needed to understand the nature of these anomalies.")

# Remove the 'kmeans_cluster' column from X_processed
X_supervised = X_processed.drop(columns=['kmeans_cluster'])

# Split the preprocessed dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_supervised, y, test_size=0.2, random_state=42)

print("Data prepared for supervised learning.")
print(f"Shape of X_supervised: {X_supervised.shape}")
print(f"Shape of X_train: {X_train.shape}")
print(f"Shape of X_test: {X_test.shape}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Shape of y_test: {y_test.shape}")

# Instantiate the Decision Tree Classifier
dt_classifier = DecisionTreeClassifier(random_state=42)

# Fit the model to the training data
dt_classifier.fit(X_train, y_train)

print("Decision Tree Classifier trained successfully.")

# Make predictions on the test set
y_pred_dt = dt_classifier.predict(X_test)

# Evaluate the model
print("\nDecision Tree Classifier Performance:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_dt):.4f}")
print("Classification Report:")
print(classification_report(y_test, y_pred_dt))

# Convert y_test to numerical for evaluation if it's not already.
y_test_numeric = y_test.apply(lambda x: 1 if x == 'anomaly' else 0)

# Get probability predictions for the positive class ('anomaly')
if 'anomaly' in dt_classifier.classes_:
    anomaly_class_index = list(dt_classifier.classes_).index('anomaly')
    y_scores_dt = dt_classifier.predict_proba(X_test)[:, anomaly_class_index]
else:
    print("Warning: 'anomaly' class not directly found in classifier.classes_. Assuming it corresponds to the second column of predict_proba.")
    y_scores_dt = dt_classifier.predict_proba(X_test)[:, 1]

# Plot Precision-Recall Curve
plt.figure(figsize=(8, 6))
disp = PrecisionRecallDisplay.from_predictions(y_test_numeric, y_scores_dt, name="Decision Tree")
disp.ax_.set_title('Precision-Recall Curve for Decision Tree Classifier')
disp.ax_.set_xlabel('Recall')
disp.ax_.set_ylabel('Precision')
plt.grid(True)
plt.show()

print("Precision-Recall Curve plotted.")

# Generate the Confusion Matrix
cm = confusion_matrix(y_test, y_pred_dt, labels=dt_classifier.classes_)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=dt_classifier.classes_)

# Plot the Confusion Matrix
plt.figure(figsize=(8, 6))
disp.plot(cmap=plt.cm.Blues, values_format='d')
disp.ax_.set_title('Confusion Matrix for Decision Tree Classifier')
plt.show()

print("Confusion Matrix plotted.")

# Instantiate the Random Forest Classifier
rf_classifier = RandomForestClassifier(random_state=42)

# Fit the model to the training data
rf_classifier.fit(X_train, y_train)

print("Random Forest Classifier trained successfully.")

# Make predictions on the test set
y_pred_rf = rf_classifier.predict(X_test)

# Evaluate the model
print("\nRandom Forest Classifier Performance:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_rf):.4f}")
print("Classification Report:")
print(classification_report(y_test, y_pred_rf))

# Convert y_test to numerical for evaluation if it's not already. Let's assume 'anomaly' is the positive class (1).
y_test_numeric_rf = y_test.apply(lambda x: 1 if x == 'anomaly' else 0)

# Get probability predictions for the positive class ('anomaly')
if 'anomaly' in rf_classifier.classes_:
    anomaly_class_index_rf = list(rf_classifier.classes_).index('anomaly')
    y_scores_rf = rf_classifier.predict_proba(X_test)[:, anomaly_class_index_rf]
else:
    print("Warning: 'anomaly' class not directly found in classifier.classes_. Assuming it corresponds to the second column of predict_proba.")
    y_scores_rf = rf_classifier.predict_proba(X_test)[:, 1]

# Plot Precision-Recall Curve
plt.figure(figsize=(8, 6))
disp_rf = PrecisionRecallDisplay.from_predictions(y_test_numeric_rf, y_scores_rf, name="Random Forest")
disp_rf.ax_.set_title('Precision-Recall Curve for Random Forest Classifier')
disp_rf.ax_.set_xlabel('Recall')
disp_rf.ax_.set_ylabel('Precision')
plt.grid(True)
plt.show()

print("Precision-Recall Curve for Random Forest plotted.")

# Generate the Confusion Matrix
cm_rf = confusion_matrix(y_test, y_pred_rf, labels=rf_classifier.classes_)
disp_rf_cm = ConfusionMatrixDisplay(confusion_matrix=cm_rf, display_labels=rf_classifier.classes_)

# Plot the Confusion Matrix
plt.figure(figsize=(8, 6))
disp_rf_cm.plot(cmap=plt.cm.Blues, values_format='d')
disp_rf_cm.ax_.set_title('Confusion Matrix for Random Forest Classifier')
plt.show()

print("Confusion Matrix for Random Forest plotted.")

# Encode target variable 'y' to numerical labels
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

# Convert numerical labels to one-hot encoding
y_categorical = to_categorical(y_encoded)

# Split the one-hot encoded target variable as well
X_train_dnn, X_test_dnn, y_train_dnn, y_test_dnn = train_test_split(
    X_supervised, y_categorical, test_size=0.2, random_state=42
)

print("TensorFlow and Keras imported. Target variable 'y' has been one-hot encoded and split.")
print(f"Shape of y_train_dnn: {y_train_dnn.shape}")
print(f"Shape of y_test_dnn: {y_test_dnn.shape}")

# Define the DNN model architecture
model = Sequential([
    Input(shape=(X_train_dnn.shape[1],)),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(y_categorical.shape[1], activation='softmax')
])

# Compile the model
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Train the model
history = model.fit(X_train_dnn, y_train_dnn, epochs=10, batch_size=32, validation_split=0.2, verbose=1)

print("Deep Neural Network model compiled and trained.")
model.summary()

# Evaluate the model on the test set
loss, accuracy = model.evaluate(X_test_dnn, y_test_dnn, verbose=0)

print("Deep Neural Network Model Performance:")
print(f"Test Loss: {loss:.4f}")
print(f"Test Accuracy: {accuracy:.4f}")

# Get probability predictions from the DNN model
y_pred_proba_dnn = model.predict(X_test_dnn)

# Convert one-hot encoded y_test_dnn back to single numerical labels for PR curve
y_test_numeric_dnn = np.argmax(y_test_dnn, axis=1)

# Find the index of the 'anomaly' class in the original label_encoder
anomaly_class_name = 'anomaly'
if anomaly_class_name in label_encoder.classes_:
    anomaly_class_idx_in_encoder = np.where(label_encoder.classes_ == anomaly_class_name)[0][0]
else:
    print("Warning: 'anomaly' class not found in label_encoder.classes_. Assuming class 0 is anomaly.")
    anomaly_class_idx_in_encoder = 0

# y_pred_proba_dnn[:, anomaly_class_idx_in_encoder] gives probabilities for 'anomaly' class
y_scores_dnn = y_pred_proba_dnn[:, anomaly_class_idx_in_encoder]

# Plot Precision-Recall Curve
plt.figure(figsize=(8, 6))
disp_dnn = PrecisionRecallDisplay.from_predictions(y_test_numeric_dnn, y_scores_dnn, name="Deep Neural Network")
disp_dnn.ax_.set_title('Precision-Recall Curve for Deep Neural Network Classifier')
disp_dnn.ax_.set_xlabel('Recall')
disp_dnn.ax_.set_ylabel('Precision')
plt.grid(True)
plt.show()

print("Precision-Recall Curve for Deep Neural Network plotted.")

# Convert one-hot encoded y_test_dnn back to single numerical labels
y_test_true_labels = np.argmax(y_test_dnn, axis=1)

# Get class predictions from the DNN model
y_pred_dnn_probabilities = model.predict(X_test_dnn)
y_pred_dnn_labels = np.argmax(y_pred_dnn_probabilities, axis=1)

# Convert numerical labels back to original string labels for display
y_test_original_labels = label_encoder.inverse_transform(y_test_true_labels)
y_pred_original_labels = label_encoder.inverse_transform(y_pred_dnn_labels)

# Generate the Confusion Matrix
class_names = label_encoder.classes_
cm_dnn = confusion_matrix(y_test_original_labels, y_pred_original_labels, labels=class_names)
disp_dnn_cm = ConfusionMatrixDisplay(confusion_matrix=cm_dnn, display_labels=class_names)

# Plot the Confusion Matrix
plt.figure(figsize=(8, 6))
disp_dnn_cm.plot(cmap=plt.cm.Blues, values_format='d')
disp_dnn_cm.ax_.set_title('Confusion Matrix for Deep Neural Network Classifier')
plt.show()

print("Confusion Matrix for Deep Neural Network plotted.")

# Get feature importances from the Random Forest model
feature_importances = rf_classifier.feature_importances_

# Create a pandas Series with feature names
feature_names = X_supervised.columns
feature_importance_series = pd.Series(feature_importances, index=feature_names)

# Sort features by importance in descending order
sorted_feature_importances = feature_importance_series.sort_values(ascending=False)

print("Top 15 Most Important Features:")
print(sorted_feature_importances.head(15))

# Select the top N (e.g., 15) most important features
top_n_features = 15
top_features = sorted_feature_importances.head(top_n_features)

# Create a bar plot to visualize these top features
plt.figure(figsize=(12, 8))
top_features.plot(kind='barh', color='skyblue')
plt.title(f'Top {top_n_features} Most Important Features for Intrusion Detection (Random Forest)')
plt.xlabel('Feature Importance Score')
plt.ylabel('Feature Name')
plt.gca().invert_yaxis()
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

print("Bar plot of top feature importances generated.")
```

## Unsupervised Learning - K-Means for Anomaly Detection

### Subtask:
Apply K-Means clustering to detect abnormal network behavior. Determine an appropriate number of clusters, fit the model, and analyze the clusters to identify potential anomalies based on cluster size or distance from centroids.


**Reasoning**:
The first step is to import the necessary libraries, `KMeans` for clustering and `matplotlib.pyplot` for plotting the elbow curve. Then, I will proceed with determining the optimal number of clusters using the elbow method as instructed.



```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Determine an appropriate number of clusters (k) using the elbow method
sum_of_squared_distances = []
K = range(1, 11) # Iterate from 1 to 10 clusters

for k in K:
    kmeans = KMeans(n_clusters=k, n_init='auto', random_state=42)
    kmeans.fit(X_processed)
    sum_of_squared_distances.append(kmeans.inertia_)

print("Inertia calculated for K values:")
for k_val, inertia_val in zip(K, sum_of_squared_distances):
    print(f"K={k_val}: Inertia={inertia_val:.2f}")
```

**Reasoning**:
Now that the inertia values have been calculated, I will plot them against the number of clusters to visualize the elbow curve, which will help in identifying an optimal number of clusters.



```python
plt.figure(figsize=(10, 6))
plt.plot(K, sum_of_squared_distances, 'bx-')
plt.xlabel('k')
plt.ylabel('Sum of squared distances (Inertia)')
plt.title('Elbow Method For Optimal k')
plt.xticks(K)
plt.grid(True)
plt.show()
```

**Reasoning**:
Based on the elbow curve (visualized from the previous step's inertia values), an optimal number of clusters appears to be around 5, where the rate of decrease in inertia starts to slow down significantly. I will now instantiate a KMeans model with 5 clusters, fit it to the preprocessed data, and then assign the cluster labels back to the dataframe to analyze the cluster distribution.



```python
optimal_k = 5 # Based on the elbow curve from the previous step

# Instantiate and fit KMeans model with the optimal number of clusters
kmeans = KMeans(n_clusters=optimal_k, n_init='auto', random_state=42)
kmeans.fit(X_processed)

# Add cluster labels to the processed DataFrame
X_processed['kmeans_cluster'] = kmeans.labels_

print(f"KMeans model fitted with {optimal_k} clusters.")
print("Distribution of samples across clusters:")
print(X_processed['kmeans_cluster'].value_counts().sort_index())
```

**Reasoning**:
Now that the clusters have been formed and their distribution printed, I will identify potentially anomalous clusters. Since anomalies are often represented by smaller clusters, I will analyze the characteristics of these small clusters and the distances of their data points from the cluster centroids to identify outliers. This step addresses the 'analyze the clusters to identify potential anomalies based on cluster size or distance from centroids' part of the subtask.



```python
from sklearn.metrics import pairwise_distances_argmin_min
import numpy as np

# Get cluster centroids
centroids = kmeans.cluster_centers_

# Calculate the distance of each sample to its assigned centroid
distances = kmeans.transform(X_processed.drop(columns=['kmeans_cluster']))
# Get the minimum distance for each point to its assigned cluster centroid
min_distances = np.min(distances, axis=1)

# Identify clusters with very few members (potential anomalies)
cluster_counts = X_processed['kmeans_cluster'].value_counts().sort_index()
small_clusters = cluster_counts[cluster_counts < 10].index.tolist() # Example threshold of < 10 members

print(f"\nSmall clusters identified (fewer than 10 members): {small_clusters}")

# Optionally, identify points that are far from their own centroids in larger clusters
# A common approach is to use a threshold (e.g., 99th percentile of distances)
threshold = np.percentile(min_distances, 99)

anomaly_indices_by_distance = X_processed.index[min_distances > threshold].tolist()

print(f"\nNumber of potential anomalies identified by distance from centroid (top 1%): {len(anomaly_indices_by_distance)}")
print("Indices of a few potential anomalies by distance:")
print(anomaly_indices_by_distance[:10]) # Print first 10 for brevity

# Combine anomalies from small clusters and anomalies by distance threshold
# For simplicity, let's just focus on reporting these two types separately
print("\nFurther analysis needed to understand the nature of these anomalies.")
```
